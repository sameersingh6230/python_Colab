```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Ravi", "Sneha", "Vikas", "Anjali"],
    "Age": [20, 21, np.nan, 22, 120, 20, 21, 19],
    "Marks": [85, 90, 88, np.nan, 95, 450, 82, 89],
    "City": ["Jaipur", "jaipur", "Delhi", "Delhi ", "Mumbai", "Mumbai", "jaipur", "Delhi"]
}

df = pd.DataFrame(data)

print("BEFORE PREPROCESSING")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df["City"] = df["City"].str.strip().str.title()

df = df[df["Age"] <= 100]
df = df[df["Marks"] <= 100]

df["Age"] = df["Age"].astype(int)
df["Marks"] = df["Marks"].astype(float)

print("\nAFTER PREPROCESSING")
print(df)

print("\nMissing Values After Preprocessing:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
df = pd.DataFrame(data)

print("STUDENT DATASET")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nAverage Marks:", df["Marks"].mean())

print("\nHighest Marks:", df["Marks"].max())

print("\nLowest Marks:", df["Marks"].min())

print("\nAverage Study Hours:", df["StudyHours"].mean())

print("\nGender Count:")
print(df["Gender"].value_counts())

print("\nStudent Results:")
print(df["Result"].value_counts())
```
